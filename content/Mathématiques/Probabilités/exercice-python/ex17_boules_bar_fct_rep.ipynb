```python
import numpy as np
import matplotlib.pyplot as plt
import math as mt
```

```python
X=np.zeros((2,4))
X[0]=np.arange(4)
for i in range(4):
    X[1][i]=mt.comb(3,i)*mt.comb(5,4-i)/mt.comb(8,4)
print(X)
plt.vlines(X[0],0,X[1])
plt.bar(X[0],X[1])
```

```python
X_cum=X.copy()
for i in range(len(X_cum[0])):
    X_cum[1][i]=X[1][:i+1].sum()
print(X_cum)
X_cum=np.insert(X_cum,0,[-2,0],axis=1)
X_cum=np.insert(X_cum,len(X_cum[0]),[5,1],axis=1)
for i in range(len(X_cum[0])-1):
    plt.plot([X_cum[0][i],X_cum[0][i+1]],[X_cum[1][i],X_cum[1][i]],'b')
print(X_cum)
plt.show()
```

```python
espe=(X[0]*X[1]).sum()
print(espe)
var=(X[0]**2*X[1]).sum()-espe**2
print(var,var**(1/2))
```
